```python
def get_application_text(receiver_name: str, course_name: str, academic_hours: str, chair_name: str, date: str, name: str) -> str:
  return f"""

                                                        {receiver_name}
                              Заява
  Прошу Вас допустити мене до платного відпрацювання пропущених навчальних
занять з дисципліни '{course_name}' у кількості {academic_hours} академічних
годин на кафедрі '{chair_name}'.

Дата: {date}                                               {name}

"""


print(get_application_text(
    input("Введіть ПІБ одержувача:"),
    input("Введіть назву дисципліни: "),
    input("Введіть кількість академічних годин: "),
    input("Введіть назву кафедри: "),
    input("Введіть дату: "),
    input("Введіть ваше ПІБ: ")
))
```
